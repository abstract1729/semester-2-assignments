```python
import torch
```

```python
x = torch.tensor([1,2,3,4], dtype = torch.float32,requires_grad =True)
y = torch.tensor([3,4,7,8], dtype = torch.float32,requires_grad =True)
z = torch.dot(x,y)**2
w = torch.sum(x*y)
```

```python
# These are the loss functions
print(type(z.grad_fn))
print(type(w.grad_fn))
```

```python
"""
z is like the loss function here. 
A single call to backward() function evaluates the gradients w.r.t the inputs to them.
In this case, it is w.r.t the inputs, that are x and y. backward() function call leads to gradient calculation w.r.t them.
"""
print("Run this cell twice to see the gradient accumulation.")
print("Before .backward() call: (Default None)")
print(x.grad)
print(y.grad)

"""
By default, PyTorch frees the buffers and the computation graph after a .backward() call to save memory.
 When set to True, the intermediate tensors and the graph structure are preserved, 
 allowing you to call .backward() multiple times on the same graph. However, you can see gradient getting accumulated
"""
z.backward(retain_graph=True) 
print("After .backward() call")
print(x.grad)
print(y.grad)

"""
To avoid gradient accumulation, use loss.zero_grad. (Only when you use optimizer)
Here, we will need to explicitly make gradients as None
"""
x.grad = None
y.grad = None
```

```python
print("Run this cell twice to see the gradient accumulation.")
print("Before .backward() call: (Default None)")
print(x.grad)
print(y.grad)

"""
By default, PyTorch frees the buffers and the computation graph after a .backward() call to save memory.
 When set to True, the intermediate tensors and the graph structure are preserved, 
 allowing you to call .backward() multiple times on the same graph. However, you can see gradient getting accumulated
"""
z.backward(retain_graph=True) 
print("\nAfter z.backward() call")
print(x.grad)
print(y.grad)


"""You can see gradient accumulation here. One of the uses of retain_graph is to be 
able to use multiple loss function on the same input"""
w.backward()
print("\nAfter w.backward() call")
print(x.grad)
print(y.grad)
```
